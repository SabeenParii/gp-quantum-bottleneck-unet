# 16 · Slim export of the evaluation table (CPU, about 1 min)
Runtime: CPU. Reads results/p2_jit_eval/eval_per_image_all.csv and writes a compressed copy with only the columns needed for the statistics (eval_slim.csv.gz). Nothing else is changed.

```
from google.colab import drive
drive.mount('/content/drive')
import pandas as pd, os
D = '/content/drive/MyDrive/quantum-placement/results/p2_jit_eval'
if not os.path.exists(D + '/eval_per_image_all.csv'):
    import glob; D = os.path.dirname(glob.glob('/content/drive/MyDrive/**/p2_jit_eval/eval_per_image_all.csv', recursive=True)[0])
df = pd.read_csv(D + '/eval_per_image_all.csv')
cols = ['set', 'model', 'fold', 'seed', 'dataset', 'roi_id', 'keep', 'dice', 'iou', 'hd95', 'dice_const', 'flip_const', 'absdp_const',
        'flip_swap', 'absdp_swap'] + [f'dice_swap{k}' for k in range(10)] + ['epochs_run', 'best_epoch']
s = df[[c for c in cols if c in df.columns]].copy()
for c in s.columns:
    if s[c].dtype == float: s[c] = s[c].round(4)
s.to_csv(D + '/eval_slim.csv.gz', index=False, compression='gzip')
print(len(s), 'rows |', round(os.path.getsize(D + '/eval_slim.csv.gz') / 1e6, 2), 'MB | written to', D)
```
