```
# 15 · Quick results table from the training summaries (CPU, about 1 minute; no GPU needed)
from google.colab import drive
drive.mount('/content/drive')
import os, glob, json
import pandas as pd
PROJECT = '/content/drive/MyDrive/quantum-placement'
rows = []
for s in ('train', 'val'):
    for f in sorted(glob.glob(f'{PROJECT}/results/p2_jit_runs_{s}/*_fold*_seed*/summary.json')):
        d = json.load(open(f))
        rows.append({k: d.get(k) for k in ('model', 'fold', 'seed', 'dice_per_image', 'iou_per_image', 'hd95_median',
                                            'dice_sever_const', 'dice_sever_swap', 'epochs_run', 'best_epoch', 'train_seconds')}
                    | {'schedule': s})
r = pd.DataFrame(rows)
r.to_csv(f'{PROJECT}/results/p2_jit_quick_table.csv', index=False)
pd.set_option('display.width', 200)
t = r.groupby(['schedule', 'model'])[['dice_per_image', 'dice_sever_const', 'dice_sever_swap', 'hd95_median', 'epochs_run']].mean().round(4)
t['n_runs'] = r.groupby(['schedule', 'model']).size()
print(t.to_string())
print('\nsaved results/p2_jit_quick_table.csv with', len(r), 'runs')
```
