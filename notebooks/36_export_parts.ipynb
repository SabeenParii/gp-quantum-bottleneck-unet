# 36 · Export the large revision-5 per-image tables in small parts (CPU, ~5 min)
Read-only for all inputs; writes only `results/p2_rev5/export/` (gzip parts of at most ~2.5 MB).

```python
from google.colab import drive
drive.mount('/content/drive', force_remount=True, timeout_ms=300000)
import os, numpy as np, pandas as pd
IN = '/content/drive/MyDrive/quantum-placement/results/p2_rev5'; OUT = f'{IN}/export'; os.makedirs(OUT, exist_ok=True)
for name in ('rev5_frozen_per_image', 'rev5_cddcesm_per_image', 'rev5_replace_per_image'):
    d = pd.read_csv(f'{IN}/{name}.csv.gz'); n = int(np.ceil(os.path.getsize(f'{IN}/{name}.csv.gz') / 2.5e6))
    for i, part in enumerate(np.array_split(np.arange(len(d)), n)):
        d.iloc[part].to_csv(f'{OUT}/{name}.part{i:02d}.csv.gz', index=False, compression='gzip', float_format='%.17g')
    print(name, len(d), 'rows ->', n, 'parts')
print(sorted(os.listdir(OUT))); print('DONE')
```
